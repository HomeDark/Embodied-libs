```python
import matplotlib.pyplot as plt
import numpy as np
#准备数据
x = np.linspace(0,2*np.pi,100)
y = np.sin(x)

#绘制曲线
plt.plot(x,y,color='blue',linestyle='-',label='Sine Wave')

#添加标签和标题
plt.title('My first Sine Curve')
plt.xlabel('X Axis')
plt.ylabel('Y Axis')
plt.legend()
plt.grid(True)

#显示图形
plt.show()
```
